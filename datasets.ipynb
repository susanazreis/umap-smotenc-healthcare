```python
import pandas as pd
import numpy as np
from sklearn.utils import resample

def create_imbalanced_dataset(df, target_col, ratio, random_state=42):
    """
    Undersample the minority class to a given ratio of its original size.
    ratio: e.g. 0.10 means keep 10% of original minority instances
    """
    majority = df[df[target_col] == 0]
    minority = df[df[target_col] == 1]
    
    n_minority_new = max(1, int(len(minority) * ratio))
    
    minority_downsampled = resample(
        minority,
        replace=False,
        n_samples=n_minority_new,
        random_state=random_state
    )
    
    return pd.concat([majority, minority_downsampled]).reset_index(drop=True)
```

```python
import pandas as pd
import os

dataset_groups = {
    'bcwd': ['final_logs/bcwd_10_log.csv', 'final_logs/bcwd_25_log.csv', 'final_logs/bcwd_50_log.csv'],
    'bcwp': ['final_logs/bcwp_10_log.csv', 'final_logs/bcwp_25_log.csv', 'final_logs/bcwp_50_log.csv'],
    'bcc':  ['final_logs/bcc_10_log.csv',  'final_logs/bcc_25_log.csv',  'final_logs/bcc_50_log.csv'],
    'pima': ['final_logs/pima_10_log.csv', 'final_logs/pima_25_log.csv', 'final_logs/pima_50_log.csv'],
    'hcv':  ['final_logs/hcv_10_log.csv',  'final_logs/hcv_25_log.csv',  'final_logs/hcv_50_log.csv'],
    'parkinsons': ['final_logs/parkinsons_5_log.csv', 'final_logs/parkinsons_10_log.csv', 'final_logs/parkinsons_15_log.csv'],
    'ilpd': ['final_logs/ilpd_5_log.csv',  'final_logs/ilpd_10_log.csv',  'final_logs/ilpd_15_log.csv'],
}

all_files = [f for files in dataset_groups.values() for f in files]

from sklearn.preprocessing import MinMaxScaler

cols = ['Mean_Knn', 'Mean_Ratio_Knn', 'Mean_Ratio_10Knn', 'mean_test_score']

y_list = []
for file in all_files:
    data = pd.read_csv(file)
    data['params'] = data['params'].astype(str)
    for col in cols:
        data[col] = pd.to_numeric(data[col], errors='coerce')

    t = data.groupby(["Generator", "Seed", "Model", "params", "fold"])[cols].mean().reset_index()
    u = t.groupby(["Generator", "Model", "Seed", "fold"])[cols].max().reset_index()
    y = t.merge(u, on=["Generator", "Model", "Seed", "fold"] + cols)[["Generator", "Model", "Seed", "fold"] + cols].drop_duplicates()
    y = y.groupby(["Generator", "Seed", "fold"])[cols].mean().reset_index()
    y.loc[y["Generator"] == "train", ["Mean_Knn", "Mean_Ratio_Knn", "Mean_Ratio_10Knn"]] = 0
    y_list.append(y)

all_scaled = []
for y in y_list:
    scaler = MinMaxScaler()
    scaled = pd.DataFrame(
        scaler.fit_transform(y[["Mean_Knn", "Mean_Ratio_Knn", "Mean_Ratio_10Knn", "mean_test_score"]]),
        index=y.index,
        columns=["Mean_Knn", "Mean_Ratio_Knn", "Mean_Ratio_10Knn", "mean_test_score"]
    )
    scaled = scaled.merge(y[["Generator", "Seed", "fold"]], left_index=True, right_index=True)
    scaled["Privacy"] = (scaled["Mean_Knn"] + scaled["Mean_Ratio_Knn"] + scaled["Mean_Ratio_10Knn"]) / 3
    scaled = scaled.rename({"mean_test_score": "ML-Utility"}, axis=1).drop(["Mean_Knn", "Mean_Ratio_Knn", "Mean_Ratio_10Knn"], axis=1)

    scaler2 = MinMaxScaler()
    y_t = pd.DataFrame(
        scaler2.fit_transform(scaled[["Privacy", "ML-Utility"]]),
        index=scaled.index, columns=["Privacy", "ML-Utility"]
    )
    scaled = y_t.merge(scaled[["Generator", "Seed", "fold"]], left_index=True, right_index=True)
    scaled["Trade-off"] = (2 * scaled["ML-Utility"] * scaled["Privacy"]) / (scaled["ML-Utility"] + scaled["Privacy"])
    all_scaled.append(scaled)

combined = pd.concat(all_scaled)

# ── Table 1 — averaged across all 21 datasets ─────────────────────────────────
print("OVERALL RESULTS — averaged across all 21 datasets")
print("=" * 60)
mean_table = combined.groupby("Generator")[["Trade-off", "Privacy", "ML-Utility"]].mean().round(3)
std_table  = combined.groupby("Generator")[["Trade-off", "Privacy", "ML-Utility"]].std().round(3)
print("\nMean:")
print(mean_table)
print("\nStd:")
print(std_table)
```

```python
import pandas as pd
import os

dataset_groups = {
    'bcwd': ['final_logs/bcwd_10_log.csv', 'final_logs/bcwd_25_log.csv', 'final_logs/bcwd_50_log.csv'],
    'bcwp': [ 'final_logs/bcwp_50_log.csv'],
    'bcc':  ['final_logs/bcc_10_log.csv',  'final_logs/bcc_25_log.csv',  'final_logs/bcc_50_log.csv'],
    'pima': [ 'final_logs/pima_25_log.csv', 'final_logs/pima_50_log.csv'],
    'hcv':  [ 'final_logs/hcv_50_log.csv'],
    'parkinsons': ['final_logs/parkinsons_5_log.csv', 'final_logs/parkinsons_10_log.csv', 'final_logs/parkinsons_15_log.csv'],
    'ilpd': ['final_logs/ilpd_5_log.csv',  'final_logs/ilpd_10_log.csv',  'final_logs/ilpd_15_log.csv'],
}

all_files = [f for files in dataset_groups.values() for f in files]

from sklearn.preprocessing import MinMaxScaler

cols = ['Mean_Knn', 'Mean_Ratio_Knn', 'Mean_Ratio_10Knn', 'mean_test_score']

y_list = []
for file in all_files:
    data = pd.read_csv(file)
    data['params'] = data['params'].astype(str)
    for col in cols:
        data[col] = pd.to_numeric(data[col], errors='coerce')

    t = data.groupby(["Generator", "Seed", "Model", "params", "fold"])[cols].mean().reset_index()
    u = t.groupby(["Generator", "Model", "Seed", "fold"])[cols].max().reset_index()
    y = t.merge(u, on=["Generator", "Model", "Seed", "fold"] + cols)[["Generator", "Model", "Seed", "fold"] + cols].drop_duplicates()
    y = y.groupby(["Generator", "Seed", "fold"])[cols].mean().reset_index()
    y.loc[y["Generator"] == "train", ["Mean_Knn", "Mean_Ratio_Knn", "Mean_Ratio_10Knn"]] = 0
    y_list.append(y)

all_scaled = []
for y in y_list:
    scaler = MinMaxScaler()
    scaled = pd.DataFrame(
        scaler.fit_transform(y[["Mean_Knn", "Mean_Ratio_Knn", "Mean_Ratio_10Knn", "mean_test_score"]]),
        index=y.index,
        columns=["Mean_Knn", "Mean_Ratio_Knn", "Mean_Ratio_10Knn", "mean_test_score"]
    )
    scaled = scaled.merge(y[["Generator", "Seed", "fold"]], left_index=True, right_index=True)
    scaled["Privacy"] = (scaled["Mean_Knn"] + scaled["Mean_Ratio_Knn"] + scaled["Mean_Ratio_10Knn"]) / 3
    scaled = scaled.rename({"mean_test_score": "ML-Utility"}, axis=1).drop(["Mean_Knn", "Mean_Ratio_Knn", "Mean_Ratio_10Knn"], axis=1)

    scaler2 = MinMaxScaler()
    y_t = pd.DataFrame(
        scaler2.fit_transform(scaled[["Privacy", "ML-Utility"]]),
        index=scaled.index, columns=["Privacy", "ML-Utility"]
    )
    scaled = y_t.merge(scaled[["Generator", "Seed", "fold"]], left_index=True, right_index=True)
    scaled["Trade-off"] = (2 * scaled["ML-Utility"] * scaled["Privacy"]) / (scaled["ML-Utility"] + scaled["Privacy"])
    all_scaled.append(scaled)

combined = pd.concat(all_scaled)

# ── Table 1 — averaged across all 21 datasets ─────────────────────────────────
print("OVERALL RESULTS — averaged across all 16 datasets")
print("=" * 60)
mean_table = combined.groupby("Generator")[["Trade-off", "Privacy", "ML-Utility"]].mean().round(3)
std_table  = combined.groupby("Generator")[["Trade-off", "Privacy", "ML-Utility"]].std().round(3)
print("\nMean:")
print(mean_table)
print("\nStd:")
print(std_table)
```
